Single Responsibility Principle

```python
print("EXERCISE 1: Single Responsibility Principle (SRP)")
import re
class User:
    """Responsibility: Store user data only"""
    def __init__(self, name, email, password):
        self.name = name
        self.email = email
        self.password = password
class EmailValidator:
    def is_valid_email(self, email):
        pattern = r'^[^@\s]+@[^@\s]+\.[^@\s]+$'
        return re.match(pattern, email) is not None
class UserRepository:
    def save_user(self, user):
        print(f" User '{user.name}' saved to database")
        print(f"   Email: {user.email}")

class EmailService:
    """Responsibility: Send emails only"""
    def send_welcome_email(self, email):
        """
        Sends welcome email to user
        Args:
            email (str): Email address to send to
        """
        print(f" Welcome email sent to {email}")
        print(f"   Subject: Welcome to our service!")
class UserService:
    """Responsibility: Orchestrate user registration using other services"""
    def __init__(self, email_validator, user_repository, email_service):
        self.email_validator = email_validator
        self.user_repository = user_repository
        self.email_service = email_service
    def register_user(self, name, email, password):
        if not self.email_validator.is_valid_email(email):
            print(f"Invalid email format: {email}\n")
            return None
        user = User(name, email, password)
        print(f"\n👤 Registering user: {name}")
        self.user_repository.save_user(user)
        self.email_service.send_welcome_email(email)

        print(f"User {name} registered successfully!\n")
        return user
print("\n TESTING EXERCISE 1:\n")
email_validator = EmailValidator()
user_repository = UserRepository()
email_service = EmailService()
user_service = UserService(email_validator, user_repository, email_service)
user1 = user_service.register_user("John Doe", "john@example.com", "password123")
user2 = user_service.register_user("Jane Smith", "invalid-email", "password456")
user3 = user_service.register_user("Bob Wilson", "bob@company.com", "password789")
```

Open/Closed Principle

```python
print("EXERCISE 2: Open/Closed Principle (OCP)")
from abc import ABC, abstractmethod
class IPaymentMethod(ABC):
    """Abstract interface for payment methods"""
    @abstractmethod
    def process(self, amount):
        """Process payment"""
        pass
class CreditCardPayment(IPaymentMethod):
    """Concrete: Credit Card Payment"""
    def process(self, amount):
        print(f" Processing credit card payment")
        print(f"   Amount: ${amount}")
        print(f"   Card: **** **** **** 1234")
        print(f"   Status:  Approved\n")
        return True
class PayPalPayment(IPaymentMethod):
    """Concrete: PayPal Payment"""
    def process(self, amount):
        print(f"  Processing PayPal payment")
        print(f"   Amount: ${amount}")
        print(f"   Account: user@example.com")
        print(f"   Status: Approved\n")
        return True
class CryptoCurrencyPayment(IPaymentMethod):
    """Concrete: Cryptocurrency Payment"""
    def process(self, amount):
        print(f"₿ Processing cryptocurrency payment")
        print(f"   Amount: ${amount}")
        print(f"   Wallet: 1A1z7agoat2YLZW51Fc792AEXJF...")
        print(f"   Status:  Confirmed\n")
        return True
class ApplePayPayment(IPaymentMethod):
    """Concrete: Apple Pay Payment (NEW - NO MODIFICATION NEEDED!)"""
    def process(self, amount):
        print(f" Processing Apple Pay payment")
        print(f"   Amount: ${amount}")
        print(f"   Device: iPhone 14 Pro")
        print(f"   Status:  Approved\n")
        return True
class GooglePayPayment(IPaymentMethod):
    """Concrete: Google Pay Payment (NEW - STILL NO MODIFICATION NEEDED!)"""
    def process(self, amount):
        print(f" Processing Google Pay payment")
        print(f"   Amount: ${amount}")
        print(f"   Device: Android Phone")
        print(f"   Status:  Approved\n")
        return True
class PaymentProcessor:
    """
    Closed for modification: Takes any IPaymentMethod
    Open for extension: Can add new payment methods anytime
    """
    def __init__(self, payment_method):
        self.payment_method = payment_method

    def process_payment(self, amount):
        """Process payment using injected payment method"""
        return self.payment_method.process(amount)
print("\n TESTING EXERCISE 2:\n")
amount = 99.99
print("--- Payment Method 1: Credit Card ---")
credit_card = CreditCardPayment()
processor = PaymentProcessor(credit_card)
processor.process_payment(amount)
print("--- Payment Method 2: PayPal ---")
paypal = PayPalPayment()
processor = PaymentProcessor(paypal)
processor.process_payment(amount)
print("--- Payment Method 3: Cryptocurrency ---")
crypto = CryptoCurrencyPayment()
processor = PaymentProcessor(crypto)
processor.process_payment(amount)
print("--- Payment Method 4: Apple Pay (NEW!) ---")
apple_pay = ApplePayPayment()
processor = PaymentProcessor(apple_pay)
processor.process_payment(amount)
print("--- Payment Method 5: Google Pay (NEW!) ---")
google_pay = GooglePayPayment()
processor = PaymentProcessor(google_pay)
processor.process_payment(amount)
```

Singleton Pattern

```python
print("EXERCISE 3: Singleton Pattern")
from datetime import datetime
class Logger:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super(Logger, cls).__new__(cls)
            print("🔧 Logger instance created!")
        return cls._instance

    def log(self, level, message):
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        print(f"[{timestamp}] {level.upper():8} | {message}")

    def info(self, message):
        self.log("INFO", message)

    def warning(self, message):
        self.log("WARNING", message)
    def error(self, message):
        self.log("ERROR", message)

    def debug(self, message):
        """Log debug level message"""
        self.log("DEBUG", message)
print("\n TESTING EXERCISE 3:\n")

print("--- Creating First Logger Instance ---\n")
logger1 = Logger()
logger1.info("Application started")
logger1.debug("Loading configuration...")
logger1.warning("Memory usage is high")

print("\n--- Creating Second Logger Instance ---\n")
logger2 = Logger()
logger2.error("Database connection failed")
logger2.info("Attempting reconnection...")

print("\n--- Verifying Singleton Property ---\n")
print(f"logger1 is logger2: {logger1 is logger2}")
print(f"logger1 == logger2: {logger1 == logger2}")
print(f"ID of logger1: {id(logger1)}")
print(f"ID of logger2: {id(logger2)}")
print("\n--- Logging from Different Parts of Application ---\n")
def process_user_data():
    """Simulated function that needs logging"""
    logger = Logger()
    logger.info("Processing user data...")
    logger.debug("Validating input...")
    logger.info("User data processed successfully")

def save_to_database():
    """Another function that needs logging"""
    logger = Logger()
    logger.info("Saving to database...")
    logger.debug("Creating connection...")
    logger.info("Data saved successfully")

print("From process_user_data():")
process_user_data()

print("\nFrom save_to_database():")
save_to_database()

```

Factory Method Pattern

```python
print("EXERCISE 4: Factory Method Pattern")
from abc import ABC, abstractmethod
class IVehicle(ABC):
    """Abstract interface for all vehicles"""
    @abstractmethod
    def drive(self):
        pass

    @abstractmethod
    def stop(self):
        pass

    @abstractmethod
    def get_capacity(self):
        pass

    @abstractmethod
    def get_type(self):
        pass
class Car(IVehicle):
    """Concrete: Car implementation"""
    def drive(self):
        print(" Car is driving...")
        print("   Speed: 120 km/h")
        print("   Fuel: Petrol")

    def stop(self):
        print(" Car stopped safely")

    def get_capacity(self):
        return 5

    def get_type(self):
        return "Car"


class Bike(IVehicle):
    """Concrete: Bike implementation"""
    def drive(self):
        print("  Bike is riding...")
        print("   Speed: 100 km/h")
        print("   Fuel: Petrol")

    def stop(self):
        print("  Bike stopped")

    def get_capacity(self):
        return 2

    def get_type(self):
        return "Bike"


class Truck(IVehicle):
    """Concrete: Truck implementation"""
    def drive(self):
        print(" Truck is moving...")
        print("   Speed: 80 km/h")
        print("   Fuel: Diesel")

    def stop(self):
        print(" Truck stopped (Engine idling)")

    def get_capacity(self):
        return 20

    def get_type(self):
        return "Truck"


class Bus(IVehicle):
    """Concrete: Bus implementation"""
    def drive(self):
        print(" Bus is driving...")
        print("   Speed: 60 km/h")
        print("   Fuel: Diesel")

    def stop(self):
        print(" Bus stopped (Passengers alighting)")

    def get_capacity(self):
        return 50

    def get_type(self):
        return "Bus"
class VehicleFactory:
    """
    Factory for creating vehicles
    Hides creation logic from client
    """
    @staticmethod
    def create_vehicle(vehicle_type):
        """
        Creates vehicle based on type
        Args:
            vehicle_type (str): Type of vehicle to create
        Returns:
            IVehicle: Vehicle instance
        Raises:
            ValueError: If unknown vehicle type
        """
        vehicle_type = vehicle_type.lower().strip()

        if vehicle_type == "car":
            return Car()
        elif vehicle_type == "bike":
            return Bike()
        elif vehicle_type == "truck":
            return Truck()
        elif vehicle_type == "bus":
            return Bus()
        else:
            raise ValueError(f" Unknown vehicle type: {vehicle_type}")


print("\n TESTING EXERCISE 4:\n")
print("--- Creating Car ---\n")
car = VehicleFactory.create_vehicle("car")
print(f"Type: {car.get_type()}")
car.drive()
print(f"Capacity: {car.get_capacity()} persons\n")
car.stop()
print("\n--- Creating Bike ---\n")
bike = VehicleFactory.create_vehicle("bike")
print(f"Type: {bike.get_type()}")
bike.drive()
print(f"Capacity: {bike.get_capacity()} persons\n")
bike.stop()
print("\n--- Creating Truck ---\n")
truck = VehicleFactory.create_vehicle("truck")
print(f"Type: {truck.get_type()}")
truck.drive()
print(f"Capacity: {truck.get_capacity()} items\n")
truck.stop()
print("\n--- Creating Bus ---\n")
bus = VehicleFactory.create_vehicle("bus")
print(f"Type: {bus.get_type()}")
bus.drive()
print(f"Capacity: {bus.get_capacity()} persons\n")
bus.stop()
print("\n--- Creating Vehicles Dynamically ---\n")
vehicle_types = ["car", "bike", "truck", "bus"]
vehicles = []

for v_type in vehicle_types:
    vehicle = VehicleFactory.create_vehicle(v_type)
    vehicles.append(vehicle)
    print(f" Created: {vehicle.get_type()}")

```

Builder Pattern

```python
print("EXERCISE 5: Builder Pattern")
class Computer:
    """Complex object with many optional components"""
    def __init__(self):
        self.cpu = None
        self.ram = None
        self.storage = None
        self.gpu = None
        self.monitor = None
        self.keyboard = None
        self.mouse = None
        self.power_supply = None
        self.case = None

    def display_configuration(self):
        """Display computer configuration"""

        print("COMPUTER CONFIGURATION")

        print(f"CPU:           {self.cpu or 'Not specified'}")
        print(f"RAM:           {self.ram or 'Not specified'}")
        print(f"Storage:       {self.storage or 'Not specified'}")
        print(f"GPU:           {self.gpu or 'Not specified'}")
        print(f"Monitor:       {self.monitor or 'Not specified'}")
        print(f"Keyboard:      {self.keyboard or 'Not specified'}")
        print(f"Mouse:         {self.mouse or 'Not specified'}")
        print(f"Power Supply:  {self.power_supply or 'Not specified'}")
        print(f"Case:          {self.case or 'Not specified'}")

class ComputerBuilder:
    def __init__(self):
        self.computer = Computer()

    def with_cpu(self, cpu):
        """Set CPU"""
        self.computer.cpu = cpu
        return self

    def with_ram(self, ram):
        """Set RAM"""
        self.computer.ram = ram
        return self

    def with_storage(self, storage):
        """Set Storage"""
        self.computer.storage = storage
        return self

    def with_gpu(self, gpu):
        """Set GPU"""
        self.computer.gpu = gpu
        return self

    def with_monitor(self, monitor):
        """Set Monitor"""
        self.computer.monitor = monitor
        return self

    def with_keyboard(self, keyboard):
        """Set Keyboard"""
        self.computer.keyboard = keyboard
        return self

    def with_mouse(self, mouse):
        """Set Mouse"""
        self.computer.mouse = mouse
        return self

    def with_power_supply(self, power_supply):
        """Set Power Supply"""
        self.computer.power_supply = power_supply
        return self

    def with_case(self, case):
        """Set Case"""
        self.computer.case = case
        return self

    def build(self):
        """Return the built computer"""
        return self.computer


print("\n TESTING EXERCISE 5:\n")

print("--- Building Gaming Computer ---\n")
gaming_pc = (ComputerBuilder()
    .with_cpu("Intel i9-13900K")
    .with_ram("32GB DDR5 6000MHz")
    .with_storage("2TB NVMe SSD")
    .with_gpu("NVIDIA RTX 4090")
    .with_monitor("4K 144Hz IPS")
    .with_keyboard("Mechanical RGB")
    .with_mouse("Gaming Mouse 16000DPI")
    .with_power_supply("1200W Modular")
    .with_case("Tempered Glass Mid-Tower")
    .build())

gaming_pc.display_configuration()

print("--- Building Budget Computer ---\n")
budget_pc = (ComputerBuilder()
    .with_cpu("AMD Ryzen 5 5500")
    .with_ram("16GB DDR4 3200MHz")
    .with_storage("256GB SSD")
    .with_case("Standard Mid-Tower")
    .build())

budget_pc.display_configuration()

print("--- Building Office Computer ---\n")
office_pc = (ComputerBuilder()
    .with_cpu("Intel i5-13600K")
    .with_ram("8GB DDR4")
    .with_storage("512GB SSD")
    .with_monitor("24 inch FHD")
    .with_keyboard("Standard Keyboard")
    .with_mouse("USB Mouse")
    .build())

office_pc.display_configuration()

print("--- Building Minimal Computer ---\n")
minimal_pc = (ComputerBuilder()
    .with_cpu("Intel Pentium")
    .with_ram("4GB DDR4")
    .with_storage("128GB SSD")
    .build())

minimal_pc.display_configuration()

print("--- Building Workstation Computer ---\n")
workstation = (ComputerBuilder()
    .with_cpu("Intel Xeon W9-3495X")
    .with_ram("256GB DDR5")
    .with_storage("4TB NVMe SSD")
    .with_gpu("NVIDIA RTX 6000 Ada")
    .with_monitor("4K 60Hz Professional")
    .build())

workstation.display_configuration()

```

Decorator Pattern

```python
print("EXERCISE 6: Decorator Pattern")
from abc import ABC, abstractmethod
class ICoffee(ABC):
    """Interface for coffee and coffee decorators"""
    @abstractmethod
    def get_description(self):
        pass
    @abstractmethod
    def get_cost(self):
        pass
class SimpleCoffee(ICoffee):
    """Base coffee without any additions"""
    def get_description(self):
        return "Simple Coffee"

    def get_cost(self):
        return 2.00
class CoffeeDecorator(ICoffee):
    """
    Base decorator for coffee
    Wraps a coffee and adds functionality
    """
    def __init__(self, coffee):
        self._coffee = coffee

    def get_description(self):
        return self._coffee.get_description()

    def get_cost(self):
        return self._coffee.get_cost()
class MilkDecorator(CoffeeDecorator):
    """Adds milk to coffee"""
    def get_description(self):
        return super().get_description() + ", Milk"

    def get_cost(self):
        return super().get_cost() + 0.50


class SugarDecorator(CoffeeDecorator):
    """Adds sugar to coffee"""
    def get_description(self):
        return super().get_description() + ", Sugar"

    def get_cost(self):
        return super().get_cost() + 0.25


class VanillaDecorator(CoffeeDecorator):
    """Adds vanilla to coffee"""
    def get_description(self):
        return super().get_description() + ", Vanilla"

    def get_cost(self):
        return super().get_cost() + 0.75


class WhippedCreamDecorator(CoffeeDecorator):
    """Adds whipped cream to coffee"""
    def get_description(self):
        return super().get_description() + ", Whipped Cream"

    def get_cost(self):
        return super().get_cost() + 0.60


class ChocolateDecorator(CoffeeDecorator):
    """Adds chocolate to coffee"""
    def get_description(self):
        return super().get_description() + ", Chocolate"

    def get_cost(self):
        return super().get_cost() + 0.55


class CaramelDecorator(CoffeeDecorator):
    """Adds caramel to coffee"""
    def get_description(self):
        return super().get_description() + ", Caramel"

    def get_cost(self):
        return super().get_cost() + 0.45
print("\n TESTING EXERCISE 6:\n")
print("--- Simple Coffee ---")
coffee = SimpleCoffee()
print(f"{coffee.get_description():40} | ${coffee.get_cost():.2f}\n")
print("--- Adding Milk ---")
coffee = MilkDecorator(SimpleCoffee())
print(f"{coffee.get_description():40} | ${coffee.get_cost():.2f}\n")
print("--- Adding Milk and Sugar ---")
coffee = SugarDecorator(MilkDecorator(SimpleCoffee()))
print(f"{coffee.get_description():40} | ${coffee.get_cost():.2f}\n")
print("--- Complex Coffee with Multiple Add-ons ---")
coffee = (VanillaDecorator(
    WhippedCreamDecorator(
        SugarDecorator(
            MilkDecorator(
                SimpleCoffee()
            )
        )
    )
))
print(f"{coffee.get_description():40} | ${coffee.get_cost():.2f}\n")
print("--- Different Coffee Combinations ---\n")

combinations = [
    ("Basic", SimpleCoffee()),

    ("Milk + Sugar", SugarDecorator(MilkDecorator(SimpleCoffee()))),

    ("Vanilla Latte", VanillaDecorator(MilkDecorator(SimpleCoffee()))),

    ("Mocha", ChocolateDecorator(MilkDecorator(SimpleCoffee()))),

    ("Caramel Latte", CaramelDecorator(MilkDecorator(SimpleCoffee()))),

    ("Luxury", WhippedCreamDecorator(
        VanillaDecorator(
            ChocolateDecorator(
                MilkDecorator(SimpleCoffee())
            )
        )
    )),
]

for name, coffee in combinations:
    description = coffee.get_description()
    cost = coffee.get_cost()
    print(f"{name:20} | {description:50} | ${cost:6.2f}")

# Test Case 6: Dynamic Decoration
print("\n--- Dynamic Decoration ---\n")
print("Building custom coffee step by step:\n")

coffee = SimpleCoffee()
print(f"Step 1: {coffee.get_description():40} | ${coffee.get_cost():.2f}")

coffee = MilkDecorator(coffee)
print(f"Step 2: {coffee.get_description():40} | ${coffee.get_cost():.2f}")

coffee = SugarDecorator(coffee)
print(f"Step 3: {coffee.get_description():40} | ${coffee.get_cost():.2f}")

coffee = VanillaDecorator(coffee)
print(f"Step 4: {coffee.get_description():40} | ${coffee.get_cost():.2f}")

coffee = WhippedCreamDecorator(coffee)
print(f"Step 5: {coffee.get_description():40} | ${coffee.get_cost():.2f}\n")

```

Adapter Pattern

```python
print("EXERCISE 7: Adapter Pattern")
from abc import ABC, abstractmethod
class IModernPaymentProcessor(ABC):
    """Modern payment interface"""
    @abstractmethod
    def execute_payment(self, amount, account_number):
        pass
class LegacyPaymentSystem:
    """Old payment system with different interface"""
    def pay(self, money, account):
        print(f"[LEGACY SYSTEM] Processing payment")
        print(f"   Money: ${money}")
        print(f"   Account: {account}")
        print(f"   Status: Transaction completed\n")
class PaymentAdapter(IModernPaymentProcessor):
    """
    Adapter to make legacy system compatible with modern interface
    Translates new interface to old interface
    """
    def __init__(self, legacy_system):
        self._legacy_system = legacy_system

    def execute_payment(self, amount, account_number):
        """
        Modern interface method
        Calls legacy system method internally
        """
        self._legacy_system.pay(amount, account_number)
class ModernPaymentProcessor:
    """Modern payment processor expecting modern interface"""
    def __init__(self, payment_method):
        self._payment_method = payment_method

    def process(self, amount, account):
        print(f"[MODERN PROCESSOR] Initiating payment")
        print(f"   Validating amount...")
        print(f"   Checking account...")
        self._payment_method.execute_payment(amount, account)
        print(f"   Recording transaction...")

print("\n TESTING EXERCISE 7:\n")

print("--- Test 1: Modern Payment System ---\n")

class CreditCardProcessor(IModernPaymentProcessor):
    """Modern implementation"""
    def execute_payment(self, amount, account_number):
        print(f"[MODERN] Processing credit card")
        print(f"   Amount: ${amount}")
        print(f"   Card: {account_number}")
        print(f"   Status:  Payment approved\n")

modern_payment = CreditCardProcessor()
processor = ModernPaymentProcessor(modern_payment)
processor.process(100.00, "4532-1234-5678-9010")
print("--- Test 2: Legacy System with Adapter ---\n")

legacy_system = LegacyPaymentSystem()
adapted_payment = PaymentAdapter(legacy_system)
processor = ModernPaymentProcessor(adapted_payment)
processor.process(150.00, "ACC-12345")
print("--- Test 3: Multiple Legacy Systems with Adapters ---\n")

class AnotherLegacyPaymentSystem:
    """Another old payment system"""
    def do_payment(self, amount, id):
        print(f"[OLD BANK SYSTEM] Processing transaction")
        print(f"   Amount: ${amount}")
        print(f"   Customer ID: {id}")
        print(f"   Status:  Success\n")


class AnotherPaymentAdapter(IModernPaymentProcessor):
    """Adapter for the other legacy system"""
    def __init__(self, legacy):
        self._legacy = legacy

    def execute_payment(self, amount, account_number):
        self._legacy.do_payment(amount, account_number)


another_legacy = AnotherLegacyPaymentSystem()
another_adapter = AnotherPaymentAdapter(another_legacy)
processor = ModernPaymentProcessor(another_adapter)
processor.process(200.00, "CUST-789")

```

Observer Pattern

```python
print("EXERCISE 8: Observer Pattern")
from abc import ABC, abstractmethod
class IStockObserver(ABC):
    """Observer interface - objects that want to be notified"""
    @abstractmethod
    def update(self, symbol, price):
        pass
class Stock:
    """
    Subject that notifies observers when price changes
    Maintains list of observers
    """
    def __init__(self, symbol, initial_price):
        self._symbol = symbol
        self._price = initial_price
        self._observers = []

    def attach(self, observer):
        """Add observer to notify list"""
        if observer not in self._observers:
            self._observers.append(observer)
            print(f" {observer.__class__.__name__} attached to {self._symbol}")

    def detach(self, observer):
        """Remove observer from notify list"""
        if observer in self._observers:
            self._observers.remove(observer)
            print(f" {observer.__class__.__name__} detached from {self._symbol}")

    def notify_observers(self):
        """Notify all observers of price change"""
        print(f"\n Notifying {len(self._observers)} observer(s)...\n")
        for observer in self._observers:
            observer.update(self._symbol, self._price)

    def set_price(self, new_price):
        """
        Update stock price and notify observers
        """
        if self._price != new_price:
            print(f"\n📊 {self._symbol} price changed: ${self._price:.2f} → ${new_price:.2f}")
            self._price = new_price
            self.notify_observers()

    def get_price(self):
        """Get current price"""
        return self._price
class MobileDisplay(IStockObserver):
    """Mobile app that displays stock prices"""
    def __init__(self, name):
        self._name = name

    def update(self, symbol, price):
        print(f"  📱 {self._name:25} | {symbol}: ${price:.2f}")


class DesktopDisplay(IStockObserver):
    """Desktop trading terminal"""
    def __init__(self, name):
        self._name = name

    def update(self, symbol, price):
        print(f"    {self._name:25} | {symbol} trading at ${price:.2f}")


class TabletDisplay(IStockObserver):
    """Tablet app display"""
    def __init__(self, name):
        self._name = name

    def update(self, symbol, price):
        print(f"   {self._name:25} | {symbol} = ${price:.2f}")


class EmailAlert(IStockObserver):
    """Email notification service"""
    def __init__(self, name):
        self._name = name

    def update(self, symbol, price):
        print(f"   {self._name:25} | Email: {symbol} now at ${price:.2f}")


class SMSAlert(IStockObserver):
    """SMS notification service"""
    def __init__(self, name):
        self._name = name

    def update(self, symbol, price):
        print(f"   {self._name:25} | SMS: {symbol} = ${price:.2f}")

print("\n TESTING EXERCISE 8:\n")
apple = Stock("AAPL", 150.00)
mobile = MobileDisplay("iPhone App")
desktop = DesktopDisplay("Trading Terminal")
tablet = TabletDisplay("iPad App")
email = EmailAlert("Email Service")
sms = SMSAlert("SMS Service")
print("--- Attaching Observers ---\n")
apple.attach(mobile)
apple.attach(desktop)
apple.attach(tablet)
apple.attach(email)
apple.attach(sms)
print("\n--- Test 1: Price Increase ---")
apple.set_price(152.50)
print("\n--- Test 2: Price Decrease ---")
apple.set_price(151.00)
print("\n--- Test 3: Detaching Mobile Observer ---\n")
apple.detach(mobile)
print("\n--- Test 4: Price Increase (After Detach) ---")
apple.set_price(155.75)

print("\n\n--- Test 5: Different Stock (Google) ---\n")
google = Stock("GOOGL", 100.00)
google.attach(mobile)
google.attach(desktop)
google.attach(email)

google.set_price(105.50)

print("\n--- Test 6: Selective Notifications ---\n")

# Create specific observers
trading_display = DesktopDisplay("Bloomberg Terminal")
trader_email = EmailAlert("Trader Email")

tesla = Stock("TSLA", 200.00)
tesla.attach(trading_display)
tesla.attach(trader_email)

print("\nInitial price:")
print(f"TSLA: ${tesla.get_price():.2f}\n")

print("After price change:")
tesla.set_price(210.25)

```

Strategy Pattern

```python
print("EXERCISE 9: Strategy Pattern")
from abc import ABC, abstractmethod
class IDiscountStrategy(ABC):
    """Interface for all discount strategies"""
    @abstractmethod
    def calculate_discount(self, original_price):
        pass

    @abstractmethod
    def get_name(self):
        pass
class StudentDiscount(IDiscountStrategy):
    """15% discount for students"""
    def calculate_discount(self, original_price):
        return original_price * 0.15

    def get_name(self):
        return "Student Discount (15%)"


class SeniorCitizenDiscount(IDiscountStrategy):
    """20% discount for senior citizens"""
    def calculate_discount(self, original_price):
        return original_price * 0.20

    def get_name(self):
        return "Senior Citizen Discount (20%)"


class RegularCustomerDiscount(IDiscountStrategy):
    """5% discount for regular customers"""
    def calculate_discount(self, original_price):
        return original_price * 0.05

    def get_name(self):
        return "Regular Customer Discount (5%)"


class BlackFridayDiscount(IDiscountStrategy):
    """50% discount for Black Friday sale"""
    def calculate_discount(self, original_price):
        return original_price * 0.50

    def get_name(self):
        return "Black Friday Super Sale (50%)"


class CorporateDiscount(IDiscountStrategy):
    """30% discount for corporate orders"""
    def calculate_discount(self, original_price):
        return original_price * 0.30

    def get_name(self):
        return "Corporate Bulk Order Discount (30%)"


class LoyaltyDiscount(IDiscountStrategy):
    """25% discount for loyal customers"""
    def calculate_discount(self, original_price):
        return original_price * 0.25

    def get_name(self):
        return "Loyalty Program Discount (25%)"
class ShoppingCart:
    """
    Shopping cart with configurable discount strategy
    Strategy can be changed at runtime
    """
    def __init__(self, discount_strategy, total_amount):
        self._discount_strategy = discount_strategy
        self._total_amount = total_amount

    def set_discount_strategy(self, strategy):
        """Change discount strategy at runtime"""
        self._discount_strategy = strategy

    def calculate_final_price(self):
        """Calculate final price with applied discount"""
        discount = self._discount_strategy.calculate_discount(self._total_amount)
        final_price = self._total_amount - discount

        print(f"Strategy Applied: {self._discount_strategy.get_name()}")
        print(f"Original Price:   ${self._total_amount:>8.2f}")
        print(f"Discount Amount:  ${discount:>8.2f}")
        print(f"Final Price:      ${final_price:>8.2f}\n")

        return final_price

print("\n📝 TESTING EXERCISE 9:\n")

total_amount = 100.00

print("--- Test 1: Student Purchase ---\n")
student_cart = ShoppingCart(StudentDiscount(), total_amount)
student_cart.calculate_final_price()

print("--- Test 2: Senior Citizen Purchase ---\n")
senior_cart = ShoppingCart(SeniorCitizenDiscount(), total_amount)
senior_cart.calculate_final_price()

print("--- Test 3: Regular Customer Purchase ---\n")
regular_cart = ShoppingCart(RegularCustomerDiscount(), total_amount)
regular_cart.calculate_final_price()

print("--- Test 4: Black Friday Sale ---\n")
sale_cart = ShoppingCart(BlackFridayDiscount(), total_amount)
sale_cart.calculate_final_price()

print("--- Test 5: Corporate Bulk Order ---\n")
corp_cart = ShoppingCart(CorporateDiscount(), total_amount)
corp_cart.calculate_final_price()

print("--- Test 6: Loyal Customer ---\n")
loyal_cart = ShoppingCart(LoyaltyDiscount(), total_amount)
loyal_cart.calculate_final_price()

print("--- Test 7: Dynamic Strategy Change ---\n")

print("Initial: Regular Customer")
cart = ShoppingCart(RegularCustomerDiscount(), total_amount)
cart.calculate_final_price()

print("Changed to: Student")
cart.set_discount_strategy(StudentDiscount())
cart.calculate_final_price()

print("Changed to: Senior Citizen")
cart.set_discount_strategy(SeniorCitizenDiscount())
cart.calculate_final_price()

print("Changed to: Black Friday")
cart.set_discount_strategy(BlackFridayDiscount())
cart.calculate_final_price()

# Test Case 8: Different Amounts
print("--- Test 8: Different Purchase Amounts ---\n")

amounts = [50, 100, 250, 500]
strategy = StudentDiscount()

print(f"Applying: {strategy.get_name()}\n")
print(f"{'Original':>15} | {'Discount':>15} | {'Final Price':>15}")
print("-" * 50)

for amount in amounts:
    cart = ShoppingCart(strategy, amount)
    discount = strategy.calculate_discount(amount)
    final = amount - discount
    print(f"${amount:>14.2f} | ${discount:>14.2f} | ${final:>14.2f}")

```

Command Pattern

```python
print("EXERCISE 10: Command Pattern")
from abc import ABC, abstractmethod
from collections import deque
class Light:
    """Light device"""
    def __init__(self):
        self._is_on = False

    def turn_on(self):
        self._is_on = True
        print("   Light is ON")

    def turn_off(self):
        self._is_on = False
        print("    Light is OFF")

    @property
    def is_on(self):
        return self._is_on


class Fan:
    """Fan device"""
    def __init__(self):
        self._is_on = False

    def start(self):
        self._is_on = True
        print("      Fan is running")

    def stop(self):
        self._is_on = False
        print("      Fan stopped")

    @property
    def is_on(self):
        return self._is_on


class TV:
    """TV device"""
    def __init__(self):
        self._is_on = False

    def turn_on(self):
        self._is_on = True
        print("     TV is ON")

    def turn_off(self):
        self._is_on = False
        print("     TV is OFF")

    @property
    def is_on(self):
        return self._is_on


class ICommand(ABC):
    """Interface for all commands"""
    @abstractmethod
    def execute(self):
        pass

    @abstractmethod
    def undo(self):
        pass


class LightOnCommand(ICommand):
    """Command to turn light on"""
    def __init__(self, light):
        self._light = light

    def execute(self):
        self._light.turn_on()

    def undo(self):
        self._light.turn_off()


class LightOffCommand(ICommand):
    """Command to turn light off"""
    def __init__(self, light):
        self._light = light

    def execute(self):
        self._light.turn_off()

    def undo(self):
        self._light.turn_on()


class FanStartCommand(ICommand):
    """Command to start fan"""
    def __init__(self, fan):
        self._fan = fan

    def execute(self):
        self._fan.start()

    def undo(self):
        self._fan.stop()


class FanStopCommand(ICommand):
    """Command to stop fan"""
    def __init__(self, fan):
        self._fan = fan

    def execute(self):
        self._fan.stop()

    def undo(self):
        self._fan.start()


class TVOnCommand(ICommand):
    """Command to turn TV on"""
    def __init__(self, tv):
        self._tv = tv

    def execute(self):
        self._tv.turn_on()

    def undo(self):
        self._tv.turn_off()


class TVOffCommand(ICommand):
    """Command to turn TV off"""
    def __init__(self, tv):
        self._tv = tv

    def execute(self):
        self._tv.turn_off()

    def undo(self):
        self._tv.turn_on()

class SmartHomeController:
    """
    Invoker that executes commands
    Maintains command history for undo/redo
    """
    def __init__(self):
        self._command_history = deque()
        self._redo_stack = deque()

    def execute_command(self, command):
        """Execute command and add to history"""
        command.execute()
        self._command_history.append(command)
        self._redo_stack.clear()

    def undo(self):
        """Undo last command"""
        if self._command_history:
            command = self._command_history.pop()
            command.undo()
            self._redo_stack.append(command)
            print("    ↶ Undo executed\n")
        else:
            print("    Nothing to undo\n")

    def redo(self):
        """Redo last undone command"""
        if self._redo_stack:
            command = self._redo_stack.pop()
            command.execute()
            self._command_history.append(command)
            print("    ↷ Redo executed\n")
        else:
            print("     Nothing to redo\n")


print("\n TESTING EXERCISE 10:\n")

controller = SmartHomeController()
light = Light()
fan = Fan()
tv = TV()

print("--- Test 1: Executing Commands ---\n")
print("Command 1: Turn Light ON")
controller.execute_command(LightOnCommand(light))

print("Command 2: Start Fan")
controller.execute_command(FanStartCommand(fan))

print("Command 3: Turn TV ON")
controller.execute_command(TVOnCommand(tv))

print("\n--- Test 2: Undoing Commands ---\n")
print("Undo Command 1:")
controller.undo()

print("Undo Command 2:")
controller.undo()

print("Undo Command 3:")
controller.undo()

print("\n--- Test 3: Redoing Commands ---\n")
print("Redo Command 1:")
controller.redo()

print("Redo Command 2:")
controller.redo()

print("Redo Command 3:")
controller.redo()

print("\n--- Test 4: Complex Sequence ---\n")

controller = SmartHomeController()

print("Executing sequence:")
print("1. Turn Light ON")
controller.execute_command(LightOnCommand(light))

print("2. Start Fan")
controller.execute_command(FanStartCommand(fan))

print("3. Turn TV ON")
controller.execute_command(TVOnCommand(tv))

print("4. Turn Light OFF")
controller.execute_command(LightOffCommand(light))

print("\nUndoing last 2 commands:")
controller.undo()
controller.undo()

print("Redoing last undone:")
controller.redo()

print("\n--- Test 5: Multiple Operations ---\n")

print("Execute: Turn Fan OFF")
controller.execute_command(FanStopCommand(fan))

print("\nCurrent state:")
print(f"  Light: {'ON' if light.is_on else 'OFF'}")
print(f"  Fan:   {'ON' if fan.is_on else 'OFF'}")
print(f"  TV:    {'ON' if tv.is_on else 'OFF'}\n")

print("Undo Fan OFF")
controller.undo()

print("\nCurrent state after undo:")
print(f"  Light: {'ON' if light.is_on else 'OFF'}")
print(f"  Fan:   {'ON' if fan.is_on else 'OFF'}")
print(f"  TV:    {'ON' if tv.is_on else 'OFF'}\n")

```

Integrated E-Commerce System

```python

print("EXERCISE 11: INTEGRATED E-COMMERCE SYSTEM")
import re
from abc import ABC, abstractmethod
from datetime import datetime
class User:
    """User model"""
    def __init__(self, name, email, password, user_type="regular"):
        self.name = name
        self.email = email
        self.password = password
        self.user_type = user_type
        self.created_at = datetime.now()

    def __repr__(self):
        return f"User({self.name}, {self.email})"


class Product:
    """Product model"""
    def __init__(self, product_id, name, price):
        self.product_id = product_id
        self.name = name
        self.price = price

    def __repr__(self):
        return f"{self.name} (${self.price})"


class Order:
    """Order model"""
    order_counter = 1000

    def __init__(self, user, products, total_amount):
        Order.order_counter += 1
        self.order_id = Order.order_counter
        self.user = user
        self.products = products
        self.total_amount = total_amount
        self.created_at = datetime.now()
        self.status = "pending"

    def __repr__(self):
        return f"Order(#{self.order_id}, {self.total_amount})"



class Logger:
    """Singleton Logger"""
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super(Logger, cls).__new__(cls)
        return cls._instance

    def log(self, level, message):
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        print(f"[{timestamp}] {level:8} | {message}")



class EmailValidator:
    """Validate email format"""
    def is_valid_email(self, email):
        pattern = r'^[^@\s]+@[^@\s]+\.[^@\s]+$'
        return re.match(pattern, email) is not None


class UserRepository:
    """Save users to database"""
    def save_user(self, user):
        logger = Logger()
        logger.log("INFO", f"Saving user: {user.name}")
        print(f"   User {user.name} saved to database\n")


class EmailService:
    """Send emails"""
    def send_welcome_email(self, user):
        logger = Logger()
        logger.log("INFO", f"Sending welcome email to {user.email}")
        print(f"   Welcome email sent to {user.email}\n")


class UserService:
    """Orchestrate user registration"""
    def __init__(self, email_validator, user_repository, email_service):
        self.email_validator = email_validator
        self.user_repository = user_repository
        self.email_service = email_service

    def register_user(self, name, email, password, user_type="regular"):
        logger = Logger()

        if not self.email_validator.is_valid_email(email):
            logger.log("ERROR", f"Invalid email: {email}")
            return None

        user = User(name, email, password, user_type)
        self.user_repository.save_user(user)
        self.email_service.send_welcome_email(user)

        logger.log("INFO", f"User {name} registered successfully")
        return user



class OrderFactory:
    """Create orders"""
    @staticmethod
    def create_order(user, products, total_amount):
        order = Order(user, products, total_amount)
        logger = Logger()
        logger.log("INFO", f"Order created: {order}")
        return order



class IPaymentStrategy(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class CreditCardPayment(IPaymentStrategy):
    def pay(self, amount):
        print(f"   Processing credit card: ${amount:.2f}")
        return True


class PayPalPayment(IPaymentStrategy):
    def pay(self, amount):
        print(f"    Processing PayPal: ${amount:.2f}")
        return True


class PaymentProcessor:
    def __init__(self, payment_strategy):
        self.payment_strategy = payment_strategy

    def process(self, amount):
        logger = Logger()
        logger.log("INFO", f"Processing payment: ${amount:.2f}")
        return self.payment_strategy.pay(amount)


class IDiscountStrategy(ABC):
    @abstractmethod
    def calculate(self, amount):
        pass

    @abstractmethod
    def name(self):
        pass


class StudentDiscount(IDiscountStrategy):
    def calculate(self, amount):
        return amount * 0.15

    def name(self):
        return "Student (15%)"


class SeniorDiscount(IDiscountStrategy):
    def calculate(self, amount):
        return amount * 0.20

    def name(self):
        return "Senior (20%)"


class RegularDiscount(IDiscountStrategy):
    def calculate(self, amount):
        return amount * 0.05

    def name(self):
        return "Regular (5%)"


class DiscountCalculator:
    def __init__(self, strategy):
        self.strategy = strategy

    def apply(self, amount):
        discount = self.strategy.calculate(amount)
        final = amount - discount
        return discount, final



class IOrderObserver(ABC):
    @abstractmethod
    def update(self, order):
        pass


class EmailNotificationObserver(IOrderObserver):
    def update(self, order):
        print(f"  📧 Order confirmation sent to {order.user.email}")


class SMSNotificationObserver(IOrderObserver):
    def update(self, order):
        print(f"  💬 SMS sent to customer")


class InventoryObserver(IOrderObserver):
    def update(self, order):
        print(f"  📦 Inventory updated for order")



class ECommerceSystem:
    """Complete E-Commerce System"""
    def __init__(self):
        # Services
        self.email_validator = EmailValidator()
        self.user_repository = UserRepository()
        self.email_service = EmailService()
        self.user_service = UserService(
            self.email_validator,
            self.user_repository,
            self.email_service
        )

        # Observers
        self.observers = []
        self.logger = Logger()

    def attach_observer(self, observer):
        """Attach observer for order notifications"""
        self.observers.append(observer)

    def notify_observers(self, order):
        """Notify all observers"""
        print("   Notifications:")
        for observer in self.observers:
            observer.update(order)
        print()

    def register_user(self, name, email, password, user_type="regular"):
        """Register new user"""
        return self.user_service.register_user(name, email, password, user_type)

    def process_order(self, user, products, total_amount,
                      discount_strategy, payment_strategy):
        """Process complete order"""
        self.logger.log("INFO", "Starting order processing")

        discount_calc = DiscountCalculator(discount_strategy)
        discount, final_amount = discount_calc.apply(total_amount)

        print(f"\n   Pricing:")
        print(f"     Original: ${total_amount:.2f}")
        print(f"     Discount ({discount_strategy.name()}): ${discount:.2f}")
        print(f"     Final: ${final_amount:.2f}\n")

        order = OrderFactory.create_order(user, products, final_amount)

        print(f"   Payment:")
        processor = PaymentProcessor(payment_strategy)
        if processor.process(final_amount):
            order.status = "completed"
            print(f"  Payment successful\n")
        self.notify_observers(order)

        return order


print("\n TESTING EXERCISE 11:\n")

system = ECommerceSystem()
system.attach_observer(EmailNotificationObserver())
system.attach_observer(SMSNotificationObserver())
system.attach_observer(InventoryObserver())

print("--- Test 1: User Registration ---\n")
john = system.register_user("John Doe", "john@example.com", "pass123", "regular")
jane = system.register_user("Jane Smith", "jane@example.com", "pass456", "student")

products = [
    Product(1, "Laptop", 1000),
    Product(2, "Mouse", 50),
    Product(3, "Keyboard", 100)
]

total = 1150

print("--- Test 2: Regular Customer Order ---\n")
print(f"📋 Processing order for {john.name}:\n")
system.process_order(
    john,
    products,
    total,
    RegularDiscount(),
    CreditCardPayment()
)

print("--- Test 3: Student Order ---\n")
print(f"Processing order for {jane.name}:\n")
system.process_order(
    jane,
    products,
    total,
    StudentDiscount(),
    PayPalPayment()
)

print("--- Test 4: Senior Customer Order ---\n")
senior = system.register_user("Bob Wilson", "bob@example.com", "pass789", "senior")
print(f"Processing order for {senior.name}:\n")
system.process_order(
    senior,
    products,
    total,
    SeniorDiscount(),
    CreditCardPayment()
)

```
