Analysis of Algorithms

```python
print("EXERCISE 1: ANALYSIS OF ALGORITHMS - TIME & SPACE COMPLEXITY")
import math

def get_first_element(arr):
    if arr:
        return arr[0]
    return None

def find_max(arr):
    if not arr:
        return None
    max_val = arr[0]
    for num in arr:
        if num > max_val:
            max_val = num
    return max_val

def bubble_sort_simple(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

def binary_search(arr, target):
    left, right = 0, len(arr) - 1

    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1

def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result

complexity_data = [
    ("O(1)", "Constant", "Accessing array element by index", "Excellent"),
    ("O(log n)", "Logarithmic", "Binary search", "Excellent"),
    ("O(n)", "Linear", "Finding max element, linear search", "Good"),
    ("O(n log n)", "Linearithmic", "Merge sort, Quick sort", "Good"),
    ("O(n²)", "Quadratic", "Bubble sort, Selection sort", "Fair"),
    ("O(n³)", "Cubic", "Nested triple loops", "Poor"),
    ("O(2^n)", "Exponential", "Recursive fibonacci", "Very Poor"),
    ("O(n!)", "Factorial", "Permutations", "Extremely Poor"),
]

print("\n" + "=" * 100)
print("BIG O NOTATION - TIME COMPLEXITY")
print("=" * 100)

print(f"\n{'Notation':<15} {'Name':<20} {'Example':<35} {'Rating':<20}")
print("-" * 90)
for notation, name, example, rating in complexity_data:
    print(f"{notation:<15} {name:<20} {example:<35} {rating:<20}")

print("\n\n" + "=" * 100)
print("PERFORMANCE COMPARISON - Operations for Different Input Sizes")
print("=" * 100)

sizes = [10, 100, 1000, 10000, 100000, 1000000]

print(f"\n{'n':<12} {'O(1)':<12} {'O(log n)':<12} {'O(n)':<12} {'O(n log n)':<12} {'O(n²)':<15}")
print("-" * 75)

for n in sizes:
    o1 = 1
    olog_n = math.log2(n)
    on = n
    on_logn = n * math.log2(n)
    on2 = n * n

    print(f"{n:<12} {o1:<12.0f} {olog_n:<12.1f} {on:<12,.0f} {on_logn:<12,.0f} {on2:<15,.0f}")

print("\n\n" + "=" * 100)
print("TESTING ALGORITHMS")
print("=" * 100)

print("\n--- Test 1: O(1) - Get First Element ---")
arr1 = [5, 2, 8, 1, 9]
print(f"Array: {arr1}")
print(f"First Element: {get_first_element(arr1)}")
print("Operations: 1 (Always)")

print("\n--- Test 2: O(n) - Find Maximum ---")
arr2 = [5, 2, 8, 1, 9, 3, 7]
print(f"Array: {arr2}")
print(f"Maximum: {find_max(arr2)}")
print(f"Operations: {len(arr2)} (One pass through array)")

print("\n--- Test 3: O(n²) - Bubble Sort ---")
arr3 = [64, 34, 25, 12, 22, 11, 90]
print(f"Unsorted: {arr3}")
sorted_arr = bubble_sort_simple(arr3.copy())
print(f"Sorted: {sorted_arr}")
print(f"Operations: {len(arr3)} x {len(arr3)} = {len(arr3)*len(arr3)} comparisons")

print("\n--- Test 4: O(log n) - Binary Search ---")
arr4 = [2, 5, 8, 12, 16, 23, 38, 45, 56, 67, 78]
target = 23
print(f"Sorted Array: {arr4}")
print(f"Search for: {target}")
index = binary_search(arr4, target)
print(f"Found at index: {index}")
print(f"Operations: log2({len(arr4)}) ≈ {math.log2(len(arr4)):.1f} comparisons")

print("\n--- Test 5: O(n log n) - Merge Sort ---")
arr5 = [64, 34, 25, 12, 22, 11, 90, 88]
print(f"Unsorted: {arr5}")
sorted_arr5 = merge_sort(arr5)
print(f"Sorted: {sorted_arr5}")
print(f"Operations: {len(arr5)} x log2({len(arr5)}) ≈ {len(arr5) * math.log2(len(arr5)):.1f}")

print("\n\n" + "=" * 100)
print("SPACE COMPLEXITY EXAMPLES")
print("=" * 100)

def space_o1():
    x = 5
    y = 10
    z = x + y
    return z

def space_on(n):
    arr = []
    for i in range(n):
        arr.append(i)
    return arr

def space_on2(n):
    matrix = []
    for i in range(n):
        row = []
        for j in range(n):
            row.append(i * j)
        matrix.append(row)
    return matrix

print("\n--- Test 1: O(1) Space ---")
result = space_o1()
print(f"Result: {result}")
print("Space used: Constant (few variables)")

print("\n--- Test 2: O(n) Space ---")
n = 5
result = space_on(n)
print(f"Result: {result}")
print(f"Space used: O({n})")

print("\n--- Test 3: O(n²) Space ---")
n = 3
result = space_on2(n)
print(f"Result: {result}")
print(f"Space used: O({n}²) = {n*n} cells")
```

Sorting Algorithms

```python
print("EXERCISE 2: SORTING ALGORITHMS")
import time
import random

def bubble_sort(arr):
    n = len(arr)
    comparisons = 0
    swaps = 0

    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            comparisons += 1
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swaps += 1
                swapped = True

        if not swapped:
            break

    return arr, comparisons, swaps

def selection_sort(arr):
    n = len(arr)
    comparisons = 0
    swaps = 0

    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            comparisons += 1
            if arr[j] < arr[min_idx]:
                min_idx = j

        if min_idx != i:
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
            swaps += 1

    return arr, comparisons, swaps

def insertion_sort(arr):
    comparisons = 0
    shifts = 0

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            comparisons += 1
            arr[j + 1] = arr[j]
            shifts += 1
            j -= 1

        arr[j + 1] = key

    return arr, comparisons, shifts

def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result

def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)

def quick_sort_inplace(arr, low=0, high=None):
    if high is None:
        high = len(arr) - 1

    if low < high:
        pi = partition(arr, low, high)
        quick_sort_inplace(arr, low, pi - 1)
        quick_sort_inplace(arr, pi + 1, high)

    return arr

def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        if arr[j] < pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

print("\nTESTING SORTING ALGORITHMS:\n")

test_arrays = [
    [64, 34, 25, 12, 22, 11, 90],
    [5, 2, 8, 1, 9],
    [1, 2, 3, 4, 5],
    [5, 4, 3, 2, 1],
]

print("--- Test 1: BUBBLE SORT ---")
for arr in test_arrays:
    sorted_arr, comp, swaps = bubble_sort(arr.copy())
    print(f"Input:  {arr}")
    print(f"Output: {sorted_arr}")
    print(f"Comparisons: {comp}, Swaps: {swaps}\n")

print("\n--- Test 2: SELECTION SORT ---")
for arr in test_arrays:
    sorted_arr, comp, swaps = selection_sort(arr.copy())
    print(f"Input:  {arr}")
    print(f"Output: {sorted_arr}")
    print(f"Comparisons: {comp}, Swaps: {swaps}\n")

print("\n--- Test 3: INSERTION SORT ---")
for arr in test_arrays:
    sorted_arr, comp, shifts = insertion_sort(arr.copy())
    print(f"Input:  {arr}")
    print(f"Output: {sorted_arr}")
    print(f"Comparisons: {comp}, Shifts: {shifts}\n")

print("\n--- Test 4: MERGE SORT ---")
for arr in test_arrays:
    sorted_arr = merge_sort(arr.copy())
    print(f"Input:  {arr}")
    print(f"Output: {sorted_arr}\n")

print("\n--- Test 5: QUICK SORT ---")
for arr in test_arrays:
    sorted_arr = quick_sort(arr.copy())
    print(f"Input:  {arr}")
    print(f"Output: {sorted_arr}\n")

print("\n\n--- Test 6: PERFORMANCE COMPARISON ---\n")

sizes = [100, 500, 1000]

print(f"{'Algorithm':<20} {'Size':<10} {'Time (ms)':<15} {'Status':<20}")
print("-" * 65)

for size in sizes:
    arr = [random.randint(1, 1000) for _ in range(size)]

    start = time.time()
    bubble_sort(arr.copy())
    bubble_time = (time.time() - start) * 1000
    print(f"{'Bubble Sort':<20} {size:<10} {bubble_time:<15.3f} {'O(n²)':<20}")

    start = time.time()
    selection_sort(arr.copy())
    selection_time = (time.time() - start) * 1000
    print(f"{'Selection Sort':<20} {size:<10} {selection_time:<15.3f} {'O(n²)':<20}")

    start = time.time()
    insertion_sort(arr.copy())
    insertion_time = (time.time() - start) * 1000
    print(f"{'Insertion Sort':<20} {size:<10} {insertion_time:<15.3f} {'O(n²)':<20}")

    start = time.time()
    merge_sort(arr.copy())
    merge_time = (time.time() - start) * 1000
    print(f"{'Merge Sort':<20} {size:<10} {merge_time:<15.3f} {'O(n log n)':<20}")

    start = time.time()
    quick_sort(arr.copy())
    quick_time = (time.time() - start) * 1000
    print(f"{'Quick Sort':<20} {size:<10} {quick_time:<15.3f} {'O(n log n)':<20}")

    print()

```
